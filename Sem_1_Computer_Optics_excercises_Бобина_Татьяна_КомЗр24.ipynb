#Задание 1
На рассеивающую линзу вдоль главной оптической оси падает параллельный пучок света диаметром 5 см. За линзой на расстоянии 20 см поставлен экран, на котором получается круглое светлое пятно диаметром 150 мм. Определить в см главное фокусное расстояние линзы.

```python
D_pad = 5  # Диаметр падающего пучка (см)
L = 20  # Расстояние от линзы до экрана (см)
D_pyatna = 15  # Диаметр пятна на экране (см)

# Расчет модуля фокусного расстояния
F_mod = (D_pad * L) / D_pyatna

# Так как линза рассеивающая, фокусное расстояние отрицательное
F = -F_mod

print(f"Решение:")
print(f"Модуль фокусного расстояния |F| = (D_пад * L) / D_пятна = ({D_pad} * {L}) / {D_pyatna} = {F_mod:.2f} см")
print(f"Ответ: F ≈ {F:.2f} см")
```

##Задание 2
Изображение предмета имеет высоту Н = 2 см. Какое фокусное расстояние F должна иметь линза, расположенная на расстоянии ƒ = 3 м от экрана, чтобы изображение данного предмета на экране имело высоту h = 0,9 м?

```python
H = 0.02   # Высота предмета (м)
h = -0.9   # Высота изображения (м) (отрицательная, т.к. действительное, перевернутое)
f = 3.0    # Расстояние от линзы до экрана (изображения) (м)

# 1. Найдем расстояние до предмета (d) из формулы увеличения
d = (f * H) / h
# Берем модуль расстояния до предмета, так как оно положительное
d_mod = abs(d)

# 2. Найдем фокусное расстояние (F) по формуле тонкой линзы
# Формула тонкой линзы: 1/F = 1/d + 1/f
F = 1 / (1/d_mod + 1/f)

print(f"Решение:")
print(f"  1. Находим расстояние до предмета d из формулы увеличения Γ = h/H = f/d:")
print(f"     d = (f * H) / h = ({f} * {H}) / ({h}) = {d:.4f} м")
print(f"     Модуль d = {d_mod:.4f} м")
print(f"  2. Находим фокусное расстояние F из формулы тонкой линзы 1/F = 1/d + 1/f:")
print(f"     1/F = 1/{d_mod:.4f} + 1/{f} = {1/d_mod:.4f} + {1/f:.4f} = {1/d_mod + 1/f:.4f}")
print(f"     F = 1 / ({1/d_mod + 1/f:.4f}) = {F:.4f} м")
print(f"Ответ: F ≈ {F * 100:.2f} см")
```

##Задание 3
Линзу с оптической силой 2,5 дптр поместили на расстоянии 0,5 м от ярко освещённого предмета. На каком расстоянии следует поместить экран, чтобы увидеть на нём чёткое изображение предмета?

```python
D = 2.5   # Оптическая сила линзы
d = 0.5   # Расстояние от линзы до предмета

# 1. Найдем фокусное расстояние линзы
F = 1 / D

# 2. Найдем расстояние до изображения (f) по формуле тонкой линзы: 1/F = 1/d + 1/f
f = 1 / (1/F - 1/d)

print(f"Решение:")
print(f"  1. Находим фокусное расстояние F = 1 / D = 1 / {D} = {F:.2f} м")
print(f"  2. Находим расстояние до экрана f из формулы тонкой линзы 1/F = 1/d + 1/f:")
print(f"     1/f = 1/F - 1/d")
print(f"     1/f = 1/{F:.2f} - 1/{d} = {1/F:.2f} - {1/d:.2f} = {1/F - 1/d:.2f}")
print(f"     f = 1 / ({1/F - 1/d:.2f}) = {f:.2f} м")
print(f"Ответ: f = {f:.2f} м")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve
import matplotlib.patches as patches
from matplotlib import animation
from IPython.display import HTML
import plotly.graph_objects as go
```

```python
class OpticalSystemSimulator:
    def __init__(self):
        self.elements = []
        self.rays = []

    def add_lens(self, position, focal_length, diameter=50, name=""):
        """Добавить тонкую линзу"""
        lens = {
            'type': 'lens',
            'position': position,
            'focal_length': focal_length,
            'diameter': diameter,
            'name': name
        }
        self.elements.append(lens)
        return lens

    def add_aperture(self, position, diameter, name=""):
        """Добавить апертуру"""
        aperture = {
            'type': 'aperture',
            'position': position,
            'diameter': diameter,
            'name': name
        }
        self.elements.append(aperture)
        return aperture

    def add_mirror(self, position, curvature, diameter=50, name=""):
        """Добавить зеркало"""
        mirror = {
            'type': 'mirror',
            'position': position,
            'curvature': curvature,
            'diameter': diameter,
            'name': name
        }
        self.elements.append(mirror)
        return mirror

    def add_ray(self, start_x, start_y, angle_deg, wavelength=550, color='red'):
        """Добавить луч"""
        ray = {
            'start': (start_x, start_y),
            'angle': np.radians(angle_deg),
            'wavelength': wavelength,
            'color': color,
            'path': []
        }
        self.rays.append(ray)
        return ray

    def trace_ray(self, ray):
        """Трассировка одного луча через систему"""
        x, y = ray['start']
        angle = ray['angle']
        path = [(x, y)]

        # Сортируем элементы по положению
        sorted_elements = sorted(self.elements, key=lambda e: e['position'])

        for element in sorted_elements:
            if element['position'] <= x:
                continue

            # Расстояние до элемента
            dx = element['position'] - x
            y_next = y + dx * np.tan(angle)

            # Проверка на попадание в апертуру
            if abs(y_next) > element.get('diameter', 100)/2:
                # Луч не проходит
                path.append((element['position'], y_next))
                break

            # Взаимодействие с оптическим элементом
            if element['type'] == 'lens':
                # Формула тонкой линзы
                f = element['focal_length']
                y_lens = y_next
                new_angle = angle - np.arctan(y_lens / f)
                angle = new_angle

            elif element['type'] == 'mirror':
                # Отражение от зеркала
                curvature = element['curvature']
                if curvature != 0:
                    # Для сферического зеркала
                    R = 1/curvature
                    normal_angle = np.arctan(y_next / R) if R != 0 else 0
                    angle = 2 * normal_angle - angle
                else:
                    # Для плоского зеркала
                    angle = -angle

            path.append((element['position'], y_next))
            x, y = element['position'], y_next

        ray['path'] = path
        return path

    def trace_all_rays(self):
        """Трассировка всех лучей"""
        for ray in self.rays:
            self.trace_ray(ray)

    def plot_system_2d(self, figsize=(15, 8)):
        """Визуализация оптической системы в 2D"""
        fig, ax = plt.subplots(figsize=figsize)

        # Рисуем оптические элементы
        for element in self.elements:
            x_pos = element['position']
            diameter = element.get('diameter', 50)

            if element['type'] == 'lens':
                # Рисуем линзу
                lens_height = diameter / 2
                ax.plot([x_pos, x_pos], [-lens_height, lens_height],
                       'b-', linewidth=3, alpha=0.7)
                ax.text(x_pos, lens_height + 5, f"f={element['focal_length']}mm",
                       ha='center', fontsize=10)

            elif element['type'] == 'aperture':
                # Рисуем апертуру
                ax.plot([x_pos, x_pos], [-diameter/2, diameter/2],
                       'k-', linewidth=2)

            elif element['type'] == 'mirror':
                # Рисуем зеркало
                curvature = element.get('curvature', 0)
                if curvature == 0:
                    # Плоское зеркало
                    ax.plot([x_pos, x_pos], [-diameter/2, diameter/2],
                           'g-', linewidth=3)
                else:
                    # Сферическое зеркало
                    R = 1/curvature
                    theta = np.arcsin(diameter/2 / abs(R))
                    arc_angles = np.linspace(-theta, theta, 100)
                    x_arc = x_pos + R * (1 - np.cos(arc_angles))
                    y_arc = R * np.sin(arc_angles)
                    ax.plot(x_arc, y_arc, 'g-', linewidth=3)

        # Рисуем лучи
        for ray in self.rays:
            path = np.array(ray['path'])
            if len(path) > 0:
                ax.plot(path[:, 0], path[:, 1],
                       color=ray['color'], linewidth=1.5, alpha=0.8)

        ax.set_xlabel('Оптическая ось (мм)')
        ax.set_ylabel('Высота (мм)')
        ax.set_title('Симуляция оптической системы')
        ax.grid(True, alpha=0.3)
        ax.axhline(0, color='k', linestyle='--', alpha=0.5)
        ax.set_xlim(0, max([e['position'] for e in self.elements]) + 50)

        plt.tight_layout()
        plt.show()

    def calculate_focus(self):
        """Расчет положения фокуса"""
        # Простой расчет для системы с линзами
        focal_points = []
        for element in self.elements:
            if element['type'] == 'lens':
                focal_points.append(element['position'] + element['focal_length'])

        return focal_points

# Создаем и настраиваем оптическую систему
simulator = OpticalSystemSimulator()

# Добавляем оптические элементы
simulator.add_lens(50, 100, 40, "Линза 1")
simulator.add_aperture(120, 30, "Апертура")
simulator.add_lens(150, 80, 40, "Линза 2")

# Добавляем лучи
for y in np.linspace(-15, 15, 5):
    simulator.add_ray(0, y, 0, color='blue')

for y in np.linspace(-12, 12, 3):
    simulator.add_ray(0, y, 2, color='red')

for y in np.linspace(-10, 10, 3):
    simulator.add_ray(0, y, -2, color='green')

# Трассируем лучи
simulator.trace_all_rays()

# Визуализируем систему
simulator.plot_system_2d()

# Расчет фокусных точек
focal_points = simulator.calculate_focus()
print(f"Расчетные фокусные точки: {focal_points}")
```

```python
def create_3d_visualization(simulator):
    fig = go.Figure()

    # Добавляем оптические элементы
    for element in simulator.elements:
        x_pos = element['position']
        diameter = element.get('diameter', 50)

        if element['type'] == 'lens':
            # Линза как цилиндр
            y = np.linspace(-diameter/2, diameter/2, 50)
            z = np.zeros_like(y)
            fig.add_trace(go.Scatter3d(
                x=[x_pos]*len(y), y=y, z=z,
                mode='lines', line=dict(color='blue', width=5),
                name=element.get('name', 'Линза')
            ))

    # Добавляем лучи
    for i, ray in enumerate(simulator.rays):
        path = np.array(ray['path'])
        if len(path) > 0:
            x = path[:, 0]
            y = path[:, 1]
            z = np.zeros_like(x)
            fig.add_trace(go.Scatter3d(
                x=x, y=y, z=z,
                mode='lines', line=dict(color=ray['color'], width=2),
                name=f'Луч {i+1}'
            ))

    fig.update_layout(
        title='3D визуализация оптической системы',
        scene=dict(
            xaxis_title='Оптическая ось (мм)',
            yaxis_title='Высота (мм)',
            zaxis_title='',
            camera=dict(eye=dict(x=1.5, y=1.5, z=1.5))
        ),
        width=800,
        height=600
    )

    return fig

# Создаем простую систему для 3D визуализации
simulator_simple = OpticalSystemSimulator()
simulator_simple.add_lens(50, 100, 40)
simulator_simple.add_ray(0, 10, 0)
simulator_simple.add_ray(0, -10, 0)
simulator_simple.trace_all_rays()

fig = create_3d_visualization(simulator_simple)
fig.show()
```
